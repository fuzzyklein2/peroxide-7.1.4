# Hello, 🌎

## Setup
```bash
$ mkdir ~/projects
$ cd ~/projects
$ mkdir hello_world
$ cd hello_world
```

## Source

### `main.rs`

```rust
//! # Hello, 🌎!
//! 
//! Entry level Rust program.

/// Print the usual greeting.
fn main() {
    println!("Hello, 🌎 !");
}
```

## Compiler

```bash
$ rustc main.rs
$ ./main
Hello, world!
```

## Cargo

### Setup

```bash
$ cargo new hello_cargo
$ cd hello_cargo
```

```bash
$ cargo run
Hello, 🌎!
```
