### Find the Clips Folder

````python
def song_folder(s: str) -> Path | None:
    """ @return The `Path` to the `songs` directory
                or `None` if it does not exist.
    """
    result = SONGS_FOLDER / s
    if result.exists(): return result
    return None
````

````python
def clips_folder(s: str) -> Path | None:
    """ @return The `Path` to the `clips` directory for the given
                song title, if it exists. Otherwise, `None`.
    """
    result = SONGS_FOLDER / s / CLIPS_DIR_NAME
    if result.exists(): return result
    return None
````

````python
clips_queue = SimpleQueue()
````

````python
class Pattern(list):
    def __init__(self, L: list):
        super().__init__(L)
        self.position = 0;
        self.length = len(self);
        self.repeats = 1;

def parse(pattern: Pattern):
    n = len(pattern)
    for j, item in enumerate(pattern):
        info(f"Parsing pattern: {pattern}")
        if type(item) is int:
            repeats_requested = item
            info(f"Repeats: {item}")
        elif type(item) is str:
            clips_queue.put(item)
            info(f"Clip: {item}")
        elif type(item) is list:
            info(f"Pattern: {item}")
            parse(item)
                        
````
